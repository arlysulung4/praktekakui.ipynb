```python
from pathlib import Path

ROOT = Path("proyek_akuisisi_data")

FOLDERS = [
    "data/raw",
    "data/interim",
    "data/processed",
    "notebooks",
    "src",
    "docs",
    "reports",
]

for folder in FOLDERS:
    path = ROOT / folder
    path.mkdir(parents=True, exist_ok=True)
    (path / ".gitkeep").touch()
    print("Dibuat:", path)

print("\nStruktur folder berhasil disiapkan.")
```
